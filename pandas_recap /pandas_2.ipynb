```python
import pandas as pd

df = pd.read_csv("../data/house_sales_in_king_county_usa.csv")
df
```

```python
df['price'].mean()
```

```python
plt.hist(df.loc[df['price'] < 2e6]['price'], bins = 100)
plt.xlabel('Price')
plt.ylabel('N houses ')
plt.show()
```

```python
df6 = df.loc[df['bedrooms'] <= 6]

plt.figure(figsize = (15,8) )
scatter = plt.scatter(df6['sqft_living'], df6['price'], s=5, c=df6['bedrooms'], cmap='inferno')

plt.title('Price as a func of living area')
plt.xlabel('Living area')
plt.ylabel('price')

char = plt.colorbar(scatter)
char.set_label('n bedrooms')
plt.show()
```

```python
df_view = df.groupby('view')['price'].mean()
plt.bar(df_view.index, df_view.values, color='purple')
plt.show()
```
