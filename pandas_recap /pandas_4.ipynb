```python
import pandas
df = pd.read_csv('../data/BCH-USDT.csv')
df
```

```python
df['date'].dtype
df['date'] = pd.to_datetime(df['date'], unit='ms')
print(df['date'].dtype)
```

```python
df
```

```python
df['weekay'] = df['date'].dt.weekday
df
```

```python
day_name_map = {
    0: 'Monday',
    1: 'Tuersday',
    2: 'Wednesday',
    3: 'Thursday',
    4: 'Friday',
    5: 'Sutarday',
    6: 'Sunday'
}

df['weekay'] = df['weekay'].map(day_name_map)
df
```

```python
df['date_diff'] = df['date'] - df['date'].shift(1)
df
```

```python
df['date_diff'].value_counts()
```

```python
df.groupby('weekay').agg({'volume':'mean'})
```
