```python
df = df.drop('Volume MA', axis=1)
df
```

```python
df = df.drop(1821)
df
```

```python
df['time'] = pd.to_datetime(df['time'], unit='s')
df
```

```python
df.loc[df['open'] == 0]
```

```python
df.set_index('time', inplace=True)
df
```

```python
df.loc['2019':'2020']
```

```python
df['low_high_avg'] = (df['low'] + df['high']) / 2
df
```

```python
df.loc[df['open'] == 0]
```

```python
df.loc[df['open'] == 0, 'open'] = df['low_high_avg']
```

```python
df.loc[df['open'] == 0]
```

```python
df.loc['2015-10-05']
```

```python
mean_prices  = df['close'].mean()
median_prices  = df['close'].median()
std_prices  = df['close'].std()

print(f"mean { round(mean_prices,2)} median {median_prices} std {std_prices}")
```

```python
df['moving_average'] = df['close'].rolling(2).mean()
df
```

```python
df['daily_returns'] = df['close'].pct_change()
df
```

```python
df['volatility'] = df['daily_returns'].rolling(30).std()
df
```

```python
df.sort_values(by='volatility', ascending=True, inplace=True)
df
```

```python
df.sort_values(by='daily_returns', ascending=True, inplace=True)
df['change'] = (df['close'] - df['open']) / df['close']
df.sort_values(by='change', ascending=True, inplace=True)
df
```
