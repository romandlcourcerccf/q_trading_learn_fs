```python
data = {
    "date": ["2022-01-01", "2022-01-02", "2022-01-03", "2022-01-04"],
    "close": [150.0, 152.0, 153.0, 154.0],
    "volume": [1000, 1200, 1300, 1250],
}
```

```python
df = pl.DataFrame(data)
df = df.with_row_index()
df
```

```python
df.select('date')
```

```python
df.filter(pl.col('close') == 160)['close', 'volume']
```

```python
print(df.index)
```

```python
df.filter((pl.col('volume') >= 1250) & (pl.col('close') > 153))
```

```python
df.filter((pl.col('volume') >= 1250) | (pl.col('close') > 153))['volume']
```
